```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import json

# Dataset preparation
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("All libraries imported successfully!")
```

```python
iris = load_iris()


X_raw = iris.data
y_raw = iris.target


feature_names = [
    'sepal_length',
    'sepal_width',
    'petal_length',
    'petal_width'
]

# class_names = iris.target_names


df = pd.DataFrame(X_raw, columns = feature_names)

df['class_id'] = y_raw
df['class_name'] = [iris.target_names[i] for i in y_raw]



print('number of samples:', len(X_raw))
print('number of features:', len(feature_names))
print('number of classes:', len(iris.target_names))

display(df.head(10))
print(df['class_name'].value_counts())
```

```python
import numpy as np

sample_ids = np.arange(len(X_raw))

print('sample ids:', sample_ids)

train_ids, temp_ids= train_test_split(
    sample_ids,
    test_size=0.4,
    random_state=42,
    # how stratify is connected with the sample ids?
    # answer: because sample_ids are just indices, stratifying by y_raw ensures that the split maintains the same class distribution as the original dataset. This is important for classification tasks to avoid bias in training and testing sets.
    stratify=y_raw 
)

print('train ids:', train_ids)  
print('temp ids:', temp_ids)

```
